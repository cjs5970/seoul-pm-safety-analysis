## **자치구별 코로나19 확진자수 일별 집계 데이터셋 전처리**

- 2020-01 ~ 2023-08 까지의 서울시 자치구별 코로나19 확진자수(일별) 데이터셋 전처리
  
- 다중 헤더(MultiIndex) 형식의 데이터셋을 읽어와 추후 분석이 용이하도록 전처리
  
- 일별로 저장된 데이터를 그룹화(`groupby`)하여 월별 데이터프레임 생성
  
- 처리한 데이터프레임(`df_grouped`)의 데이터 점검, CSV 저장 (utf-8-sig)

#### **결과물: 서울시_자치구별_월별_코로나19_확진자수.csv**

  #### **목표 데이터 설명**
  * **연월** - (YYYY-MM) 형식의 날짜입니다.
  * **자치구** - 서울시 내 25개 구의 이름입니다.
  * **월별_확진자수** - 연월-자치구에 해당하는 월별 확진자수입니다.

```python
import pandas as pd
import numpy as np
import os
```

### 데이터셋 불러오기

```python
# 원본 데이터셋 BASE_PATH 경로 정의
BASE_PATH = '../../original-datasets/서울시 공공데이터/서울시 자치구별 코로나 확진자, 사망자 현황(일자별)_2020~2023(~8월)'

# 서울시 자치구별 코로나19 확진자, 사망자 현황(일자별) 데이터 불러오기
# 컬럼명이 2행으로 되어 있으므로 header=[0, 1]로 지정하여 멀티인덱스 컬럼으로 읽어옴
df_covid = pd.read_excel(os.path.join(BASE_PATH, '서울시_자치구별_코로나 확진자, 사망자 현황(일자별).xlsx'), header=[0, 1])

print(df_covid.info())
df_covid.head()
```

### 데이터셋 전처리

#### '구분' 컬럼명이 2개의 행을 연속으로 차지하고 있어, 하위 컬럼명이 지정되지 않았습니다(0_level_1).
#### 데이터를 쉽게 다루기 위해 날짜를 인덱스로 설정합니다.

```python
# 날짜를 인덱스로 설정
df_covid = df_covid.set_index(('구분', 'Unnamed: 0_level_1'))
df_covid.index.name = '날짜'

df_covid.head()
```

#### 분석에 필요하지 않은 컬럼은 제거합니다.

```python
# 분석에 불필요한 열(총합계, 기타, 타시도) 제거
cols_to_drop = [col for col in ['총합계', '기타', '타시도'] if col in df_covid.columns.get_level_values(0)]
df_covid = df_covid.drop(columns=cols_to_drop, level=0)

df_covid.head()
```

#### Stack 메소드로 멀티인덱스를 해제합니다.

```python
# 데이터프레임을 긴 형식으로 변환 (stack)
# 멀티인덱스 컬럼에서 첫 번째 레벨(확진, 사망)을 유지하고, 두 번째 레벨(자치구)을 행으로 변환
df_long = df_covid.stack(level=0).reset_index()

# 컬럼명 정리
df_long.rename(columns={'level_1': '자치구', '확진': '일별_확진', '사망': '일별_사망'}, inplace=True)

df_long
```

#### 필요하지 않은 행은 제거하고, '연월' 열을 새로 추가합니다.

```python
# '날짜' 컬럼에 들어있는 '총합계' 행 제거
df_long = df_long[df_long['날짜'] != '총합계'].copy()

# 날짜 컬럼을 datetime 타입으로 변환
df_long['날짜'] = pd.to_datetime(df_long['날짜'], errors='coerce')
df_long = df_long.dropna(subset=['날짜']) # 날짜 변환에 실패한 행 제거

# 보기 쉽게 '연-월(YYYY-MM)' 형태로만 잘라서 새로운 컬럼 만들기
df_long['연월'] = df_long['날짜'].dt.to_period('M')

df_long
```

#### (연월, 자치구)를 기준으로, 일별 확진자수의 합계를 계산하여 월별 확진자수를 구합니다. 

```python
# 연월 / 자치구 단위로 확진자 수 월별 합산 집계 (groupby)
df_grouped = df_long.groupby(['연월', '자치구'])['일별_확진'].sum().reset_index()
df_grouped.rename(columns={'일별_확진': '월별_확진자수'}, inplace=True)

df_grouped
```

#### 결측치를 확인하고, 최종데이터셋을 CSV 파일로 저장합니다.

```python
# 결측치 확인
print(df_grouped.isnull().sum())
# 최종 데이터 크기 출력
print("최종 데이터 크기: ", df_grouped.shape)
```

```python
# 결과를 CSV 파일로 저장
df_grouped.to_csv('서울시_자치구별_월별_코로나19_확진자수.csv', index=False, encoding='utf-8-sig')
```
