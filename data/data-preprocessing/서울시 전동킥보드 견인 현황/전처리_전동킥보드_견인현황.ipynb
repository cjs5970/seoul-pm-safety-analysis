```python
import pandas as pd
import numpy as np
```

#### 필요 컬럼명 정의, 출력용 함수 정의

```python
# 필요한 컬럼명 정의
cols=['신고일', '구정보', '주소', '유형', '조치일']

# 편의용 함수 정의
def my_display(df, num_rows=5):
    display(df.head(num_rows))
    display(df.tail(num_rows))
    print(df.dtypes)
```

## 데이터셋 전처리

### 필요한 함수를 작성하여, 결측치, 데이터타입 등을 처리합니다.

```python
# 데이터프레임의 결측치 조회 함수 정의
def check_missing_values(df, df_name):
    print(f"{df_name} 결측치 개수:")
    print(df.isnull().sum())
```

```python
# 데이터 전처리 함수 정의
def preprocess_data(df):
    # 결측치가 존재하는 행 제거
    df = df.dropna(subset=['신고일', '구정보', '유형', '조치일'])

    # '신고일'과 '조치일'을 datetime 형식으로 변환
    df['신고일'] = pd.to_datetime(df['신고일'], errors='coerce')
    df['조치일'] = pd.to_datetime(df['조치일'], errors='coerce')
    
    # '구정보'에서 구 이름만 추출 (예: '강남구'에서 '강남구' 추출)
    df['구정보'] = df['구정보'].str.extract(r'(\w+구)')

    # 신고월 파생변수 생성
    df['신고월'] = df['신고일'].dt.to_period('M')
    
    return df
```

```python
# 데이터 전처리
df_20230831 = preprocess_data(df_20230831)
df_20231231 = preprocess_data(df_20231231)
df_20240630 = preprocess_data(df_20240630)
df_20241231 = preprocess_data(df_20241231)
df_20250630 = preprocess_data(df_20250630)
df_20251231 = preprocess_data(df_20251231)

# 각 데이터프레임들의 결측치 확인
print(check_missing_values(df_20230831, "2023-08-31")) # 2021년 7월부터 2023년 8월까지의 데이터는 누적되어 업데이트
print(check_missing_values(df_20231231, "2023-12-31"))
print(check_missing_values(df_20240630, "2024-06-30"))
print(check_missing_values(df_20241231, "2024-12-31"))
print(check_missing_values(df_20250630, "2025-06-30"))
print(check_missing_values(df_20251231, "2025-12-31"))
```

### 결측치, 데이터타입, 신고월 파생변수 추가가 성공적으로 완료
### 해당 데이터프레임들을 하나로 병합합니다.

```python
# 전처리된 데이터프레임들을 하나로 병합
df_merged = pd.concat(
    [df_20230831, df_20231231, 
     df_20240630, df_20241231, 
     df_20250630, df_20251231], 
     ignore_index=True)

# 병합된 데이터프레임 정보 출력
print(df_merged.info())
display(df_merged.head(3))
display(df_merged.tail(3))
```

### 병합된 결과 데이터프레임 파일 저장(CSV) 및 정상적으로 저장되었는지 확인

```python
# 병합된 데이터프레임 파일 저장
# 저장 시, 인코딩은 'utf-8-sig'로 지정
df_merged.to_csv('서울특별시_전동킥보드_견인_현황_202107~202512.csv', index=False, encoding='utf-8-sig')
```

```python
# 저장한 파일 정상적으로 저장되었는지 확인
# parse_dates 옵션을 사용하여 '신고일'과 '조치일'을 datetime 형식으로 읽어오기
df_merged_check = pd.read_csv('서울특별시_전동킥보드_견인_현황_202107~202512.csv', 
                              encoding='utf-8-sig', parse_dates=['신고일', '조치일'])
print(df_merged_check.info())
display(df_merged_check.head(3))
display(df_merged_check.tail(3))
```
